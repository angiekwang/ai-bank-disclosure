```python
import pandas as pd
from pathlib import Path
```

```python
DIRECTORY = Path.cwd().resolve().parents[0]
bank_list = pd.read_csv(DIRECTORY / "base_data" / "wrds_raw_banks.csv")
```

```python
# Standardize  data types
bank_list["at"] = bank_list["at"].astype(float)
bank_list["gvkey"] = (bank_list["gvkey"]
                      .astype(str)
                      .str.strip())
bank_list["cik"] = (bank_list["cik"]
                    .astype(str)
                    .str.strip()
                    .str.replace(r"\.$", "", regex=True)
                    .str.zfill(10)) 

print(f"Number of banks: {len(bank_list)}")

# In case: Remove duplicate banks by CIK, keeping the earliest data date
DATE_COL = "datadate"
bank_list[DATE_COL] = pd.to_datetime(bank_list[DATE_COL])
bank_list = bank_list.sort_values(by=[DATE_COL])
bank_list = bank_list.drop_duplicates("cik", keep = 'first')

print(f"Number of banks after removing duplicates: {len(bank_list)}")
```

```python
# Filter consolidated assets greater than 2 billion
bank_list = bank_list[bank_list["at"] > 2000]

print(f"Number of banks after filtering for size: {len(bank_list)}")

file_path = DIRECTORY / "base_data" / "clean_bank_list.csv"
bank_list.to_csv(file_path, index=False)
print(bank_list)
```
